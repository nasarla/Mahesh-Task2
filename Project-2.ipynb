# Outlier Detection (IQR Method)

```python
Q1 = df["Age"].quantile(0.25)

Q3 = df["Age"].quantile(0.75)

IQR = Q3-Q1

lower = Q1-1.5*IQR

upper = Q3+1.5*IQR

outliers_age = df[(df["Age"]<lower)|(df["Age"]>upper)]

print("Age Outliers =",len(outliers_age))
```

```python
Q1 = df["Fare"].quantile(0.25)

Q3 = df["Fare"].quantile(0.75)

IQR = Q3-Q1

lower = Q1-1.5*IQR

upper = Q3+1.5*IQR

outliers_fare = df[(df["Fare"]<lower)|(df["Fare"]>upper)]

print("Fare Outliers =",len(outliers_fare))
```

```python

df["Fare"] = np.where(df["Fare"] > upper, upper, df["Fare"])

df["Fare"] = np.where(df["Fare"] < lower, lower, df["Fare"])
```

```python
outliers_after = df[(df["Fare"] < lower) | (df["Fare"] > upper)]

print("Fare Outliers After:", len(outliers_after))
```

# utlier Decision
Age outliers were retained because extremely young and elderly passengers are valid observations.
Fare outliers were capped using the IQR upper bound to reduce the influence of extremely high ticket prices while preserving all records.
